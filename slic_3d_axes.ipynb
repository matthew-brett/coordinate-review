```python
import transpose_invariance as tpi

import skimage as ski
import skimage.segmentation as ss
```

```python
imgs = tpi.get_3d_images()
img = imgs[0]
```

```python
img.shape
```

```python
img = imgs[0]
func = lambda img : ss.slic(img, channel_axis=None)
ws_orig = func(img)
```

```python
ws_rolled = tpi.rolled_proc(img, (2, 1, 0), func)
tpi.assert_labels_equivalent(ws_rolled, ws_orig)
```

```python
# Cell images as transpose invariant
tpi.assert_all_orders(imgs[:2], func)
```

```python
# Brain image is not.
brain = imgs[-1]
orig = func(brain)
ws_rolled = tpi.rolled_proc(brain, (2, 1, 0), func)
# This raises an AssertionError
tpi.assert_labels_equivalent(ws_rolled, orig)
```

```python
# It is invariant with some added noise to make values unique.
brain_fog = tpi.without_ties(brain)
orig = func(brain_fog)
ws_rolled = tpi.rolled_proc(brain_fog, (2, 1, 0), func)
tpi.assert_labels_equivalent(ws_rolled, orig)
```
