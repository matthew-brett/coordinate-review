The `rotate` docstring asserts that rotation is counter-clockwise.  Let's check.

```python
import matplotlib.pyplot as plt
import skimage as ski
```

```python
img = ski.data.camera()
plt.imshow(img, cmap='gray')
```

```python
r_img = ski.transform.rotate(img, 20)
plt.imshow(r_img, cmap='gray')
```
